## **Menggabungkan Data**

Proses atau kegiatan menyatukan data dari berbagai sumber atau tabel berdasarkan hubungan tertentu agar menjadi satu informasi utuh, dengan tujuan memudahkan analisis, pengambilan keputusan, dan pemahaman data secara menyeluruh.

Dua buah data atau lebih bisa digabungkan **asalkan memiliki kolom kunci (key) yang sama atau saling berhubungan**.

Cara menggabungkan data di pandas bisa dilakukan dengan mengeksekusi perintah berikut

```
DataFrame1.merge(DataFrame2, left_on = [key_df1], right_on = [key_df2], how)
```

<br>Dengan ketentuan :     
* **DataFrame1** adalah dataframe disebelah kiri
* **DataFrame2** adalah dataframe disebelah kanan
* **key_df1** adalah kunci pada DataFrame1 yang akan digabungkan
* **key_df2** adalah kunci pada DataFrame2 yang akan digabungkan  
* **how** adalah cara penggabungan yakni bisa ‘left’, ‘right’, ‘outer’, ‘inner’, ‘cross‘ secara default adalah ‘inner’


<br>docs : *https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.merge.html*

### **Understanding Different Join Types**

When merging DataFrames, the `how` parameter in the `.merge()` function dictates how rows are combined based on matching keys. Let's explore the three common types:

1.  **Inner Join (`how='inner'`)**
    *   **What it does:** Returns only the rows that have matching values in *both* DataFrames. It's like finding the intersection of the two DataFrames based on the specified keys.
    *   **In your code (cell `bDI8JCHkmCte`):**
        ```python
        pizza_transaction = orders.merge(order_details, left_on=['order_id'], right_on=['order_id'], how='inner')
        ```
        This code combines `orders` and `order_details`. If an `order_id` exists in `orders` but not in `order_details` (or vice-versa), that row will *not* be included in `pizza_transaction`.

2.  **Left Join (`how='left'`)**
    *   **What it does:** Returns all rows from the *left* DataFrame, and the matching rows from the *right* DataFrame. If there's no match in the right DataFrame, the columns from the right DataFrame will be filled with `NaN` (Not a Number).
    *   **Example below:** We'll see how this preserves all `order_id`s from the `orders` DataFrame, even if they don't have corresponding details.

3.  **Right Join (`how='right'`)**
    *   **What it does:** Returns all rows from the *right* DataFrame, and the matching rows from the *left* DataFrame. If there's no match in the left DataFrame, the columns from the left DataFrame will be filled with `NaN`.
    *   **In your code (cell `gjljDw9teFwx`):**
        ```python
        pizza_transaction = orders.merge(order_details, left_on=['order_id'], right_on=['order_id'], how='right')
        ```
        This code prioritizes `order_details`. If an `order_id` exists in `order_details` but not in `orders` (which shouldn't happen if `order_details` always refers to an `order_id` in `orders`), that row would still be included, and `orders` columns would show `NaN`.

````python
# Misal kalau pakai left join
left_join_pizza_transaction = orders.merge(order_details, left_on=['order_id'], right_on=['order_id'], how='left')

# Tampilkan sampel datanya
display(left_join_pizza_transaction.head())

# Tampilkan jumlah baris untuk perbandingan
print(f"Number of rows with Inner Join: {pizza_transaction.shape[0]}")
print(f"Number of rows with Left Join: {left_join_pizza_transaction.shape[0]}")
print(f"Number of rows with Right Join (from previous execution): {pizza_transaction.shape[0]}")
````

#### **Summary of Differences**

*   **Inner Join:** Only records where the join key exists in *both* DataFrames.
*   **Left Join:** All records from the *left* DataFrame, plus matching records from the *right* DataFrame. If no match on the right, `NaN`s are filled.
*   **Right Join:** All records from the *right* DataFrame, plus matching records from the *left* DataFrame. If no match on the left, `NaN`s are filled.

Choosing the correct join type depends entirely on what you want to achieve with your data. For instance, if you want to ensure you don't lose any order information from your `orders` DataFrame, a `left join` would be appropriate.

##**Aggregation**

Agregasi data merupakan serangkaian proses yang dilakukan untuk melakukan pengumpulan data dan dinyatakan dalam bentuk ringkasan sebagai contoh menghitung total yang dikelompokkan berdasarkan kategori tertentu, dsb.

Untuk melakukan agregasi gunakan sintaks berikut

<p align="center"><code><b>DataFrame.groupby(by, as_index).agg(kolom_hasil_aggregasi = (kolom_dihitung, jenis_agregasi))</b></code></p>

Jenis agregasi :
* sum = Menghitung total data
* mean = Menghitung rataan data
* max = Menghitung nilai maksimum data
* min = Menghitung nilai minimum data
* nunique = Menghitung jumlah data unik data
* count = Menghitung banyak data

<br>docs : *https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.groupby.html*

### 👨‍💻 Task 7

Pada DataFrame `pizza_transaction`, lakukan perhitungan :      
* Total kuantitas
* Total amount (price)
* Total profit
* Total komplain

untuk setiap harinya pada kategori dan nama pizza terjual. Simpan hasil perhitungan kedalam dataframe `result`

````python
# Lakukan perhitungan agregasi
result = pizza_transaction.groupby(['order_date', 'pizza_name', 'pizza_category'], as_index = False).agg(
    total_quantity = ('quantity', 'sum'),
    total_amount = ('total_price', 'sum'),
    total_profit = ('total_profit', 'sum'),
    total_complaint = ('is_complain', 'sum'),
    total_transaction = ('order_id', 'nunique')
)

# Tampilkan hasilnya
display(result)
````
