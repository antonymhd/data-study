Jenis Database :
<ul>
<li><b>Relational Database</b></li>
Database yang menyimpan data dalam bentuk tabel (baris & kolom) dan saling berelasi.

<li><b>Non-Relational Database</b></li>
Database dengan struktur fleksibel, tidak berbasis tabel.<br><br>

<img src="https://docs.cloud.google.com/static/bigquery/images/gc-bigquery-ready.png" width="40%">

Google BigQuery adalah tools database yang berfungsi sebagai data warehouse di cloud untuk menyimpan dan menganalisis data dalam jumlah sangat besar menggunakan SQL.

Dari sisi database, BigQuery memiliki karakteristik berikut:
* Menyimpan data dalam bentuk tabel seperti database pada umumnya (memiliki dataset, tabel, kolom, dan tipe data).

* Dioptimalkan untuk analisis data (OLAP), bukan untuk transaksi harian seperti database aplikasi.
* Dikelola sepenuhnya di cloud, sehingga pengguna tidak perlu mengatur server, storage, atau scaling.
* Mampu memproses data sangat besar (hingga terabyte atau petabyte) dengan query yang tetap cepat.

## **Menyaring Nilai Berdasarkan Kondisi Tertentu**

Untuk melakukan penyaringan nilai berdasarkan kondisi tertentu dapat menggunakan <b>Comparison Operators</b>. Operator perbandingan (<i>comparison operators</i>) dalam SQL digunakan untuk membandingkan nilai-nilai dan menentukan bagaimana data memenuhi kondisi tertentu. Operator pembanding pada SQL disimbolkan dengan beberapa tanda berikut
<br><br>
<table border="1">
    <tr>
        <th>Tanda</th>
        <th>Deskripsi</th>
        <th>Contoh</th>
    </tr>
    <tr>
        <td>=</td>
        <td>Sama dengan (<i>Equals</i>)</td>
        <td>A = B</td>
    </tr>
    <tr>
        <td><></td>
        <td>Tidak Sama dengan (<i>Not Equals</i>)</td>
        <td>A <> B</td>
    </tr>
    <tr>
        <td><</td>
        <td>Kurang dari (<i>Less than</i>)</td>
        <td>A < B</td>
    </tr>
    <tr>
        <td>></td>
        <td>Lebih dari (<i>More than</i>)</td>
        <td>A > B</td>
    </tr>
    <tr>
        <td><=</td>
        <td>Kurang dari sama dengan (<i>Less than or Equals To</i>)</td>
        <td>A <= B</td>
    </tr>
    <tr>
        <td>>=</td>
        <td>Lebih dari sama dengan (<i>More than or Equals To</i>)</td>
        <td>A >= B</td>
    </tr>
</table>

<br>Kemudian gunakan operator pembanding tersebut kedalam sebuah kondisi <br><br>
```

WHERE kondisi;
```

## **Menyaring Nilai Berdasarkan Beberapa Kondisi**

Operator logika (<i>logical operator</i>) adalah simbol atau kata kunci yang digunakan untuk menghubungkan dua atau lebih ekspresi logis dan menghasilkan nilai logis baru berdasarkan operasi tersebut. Operator logika pada SQL disimbolkan dengan beberapa tanda berikut

# **Menggabungkan Tabel**

Data biasanya disimpan di berbagai tabel atau sistem yang berbeda pada RDBMS (Relational Database Management System). JOIN memungkinkan pengguna untuk menggabungkan data dari berbagai sumber untuk mendapatkan gambaran yang lebih lengkap. Memahami query JOIN memungkinkan seorang pengguna untuk mengintegrasikan dan menganalisis data secara efektif, sehingga dapat meningkatkan kualitas analisis dan mendukung pengambilan keputusan berbasis data.

Operasi JOIN standar dapat dilakukan jika terdapat kunci pada tiap tabel yang saling terhubung. Operasi join dasar yang paling umum digunakan ada 4 yakni :
<ul>
    <img src="https://media.licdn.com/dms/image/v2/D5622AQENtpe3uskDew/feedshare-shrink_800/feedshare-shrink_800/0/1702313123963?e=2147483647&v=beta&t=8moZqvVqoGRwHsTHZfRDCK-rcjCqb3DHqL3v2Bywh1Y" width="40%">
    <li><b>LEFT JOIN</b></li>
    LEFT JOIN adalah jenis operasi penggabungan baris dari dua tabel, tetapi hanya mengembalikan semua baris dari tabel kiri (tabel pertama) dan baris yang cocok dari tabel kanan (tabel kedua). Jika tidak ada kecocokan di tabel kanan, maka nilai-nilai untuk kolom dari tabel kanan akan diisi dengan NULL.
    <br>
    <li><b>RIGHT JOIN</b></li>
    RIGHT JOIN adalah jenis operasi penggabungan baris dari dua tabel, tetapi hanya mengembalikan semua baris dari tabel kanan (tabel kedua) dan baris yang cocok dari tabel kiri (tabel pertama). Jika tidak ada kecocokan di tabel kiri, maka nilai-nilai untuk kolom dari tabel kiri akan diisi dengan NULL.
    <br>
    <li><b>INNER JOIN</b></li>
    INNER JOIN adalah jenis operasi penggabungan yang hanya mengembalikan baris yang memiliki kecocokan di kedua tabel berdasarkan kriteria tertentu.
    <br>
    <li><b>FULL JOIN</b></li>
    FULL JOIN adalah jenis operasi penggabungan semua baris dari kedua tabel, baik yang memiliki kecocokan maupun yang tidak. Dengan demikian, hasilnya akan mencakup semua data dari tabel kiri dan tabel kanan.
    <br>
</ul>

### 📝 **Task** : ***Laporan Ringkasan Penjualan Harian***

Manajemen membutuhkan laporan ringkasan harian untuk menganalisis performa penjualan pizza. Buatlah sebuah laporan yang menampilkan metrik-metrik berikut ini per tanggal:

1.  **Tanggal Pesanan**
2.  **Total Transaksi**: Jumlah pesanan yang terjadi pada tanggal tersebut.
3.  **Total Pizza Terjual**: Total kuantitas pizza yang terjual.
4.  **Total Pendapatan (Revenue)**: Total nilai penjualan pizza.
5.  **Total Keuntungan (Profit)**: Total keuntungan yang didapatkan dari penjualan pizza (dengan mempertimbangkan harga jual dan biaya produksi).
6.  **Total Komplain**: Jumlah komplain yang tercatat pada tanggal tersebut.

Pastikan untuk menggabungkan data dari berbagai tabel yang ada, untuk mendapatkan semua informasi yang dibutuhkan. Perhatikan juga konversi tipe data yang mungkin diperlukan untuk perhitungan.

Note : *Untuk mendapatkan data sesuai kriteria pelajari konsep `Konversi Tipe Data`, `Fungsi JOIN`, `Agregasi` dan juga `Pengurutan`*


````python
%%bigquery --project {project_id} --verbose

SELECT
    PARSE_DATE('%m/%d/%Y', o.order_date) AS tanggal_pesanan,
    COUNT(DISTINCT o.order_id) AS total_transaksi,
    SUM(od.quantity) AS total_pizza_terjual,
    SUM(od.quantity * CAST(REPLACE(p.price, 'IDR', '') AS NUMERIC)) AS total_pendapatan,
    SUM(od.quantity * (CAST(REPLACE(p.price, 'IDR', '') AS NUMERIC) - CAST(REPLACE(p.production_cost, 'IDR', '') AS NUMERIC))) AS total_keuntungan,
    SUM(CASE WHEN o.is_complain = 1 THEN 1 ELSE 0 END) AS total_komplain
FROM
    pizza_transaction.trx_order AS o
JOIN
    pizza_transaction.trx_order_detail AS od ON o.order_id = od.order_id
JOIN
    pizza_transaction.mst_pizza AS p ON od.pizza_id = p.pizza_id
GROUP BY
    tanggal_pesanan
ORDER BY
    tanggal_pesanan
````

### **Analisis: Rata-rata Pizza per Pesanan**

Untuk mengetahui berapa banyak pizza yang biasanya ada dalam satu pesanan, kita bisa menghitung rata-rata kuantitas pizza per `order_id`.

````python
%%bigquery --project {project_id} --verbose

SELECT
    ROUND(AVG(total_pizza_per_order), 2) AS rata_rata_pizza_per_pesanan
FROM (
    SELECT
        o.order_id,
        SUM(od.quantity) AS total_pizza_per_order
    FROM
        pizza_transaction.trx_order AS o
    JOIN
        pizza_transaction.trx_order_detail AS od ON o.order_id = od.order_id
    GROUP BY
        o.order_id
)
````

### **Analisis: Produk Pizza Terlaris (Best Seller)**

Untuk menemukan produk pizza terlaris, kita akan menghitung total kuantitas yang terjual untuk setiap tipe pizza. Ini melibatkan penggabungan tabel detail pesanan (`trx_order_detail`), informasi pizza (`mst_pizza`), dan tipe pizza (`mst_pizza_type`).

````python
%%bigquery --project {project_id} --verbose

SELECT
    pt.name AS nama_pizza,
    SUM(od.quantity) AS total_kuantitas_terjual
FROM
    pizza_transaction.trx_order_detail AS od
JOIN
    pizza_transaction.mst_pizza AS p ON od.pizza_id = p.pizza_id
JOIN
    pizza_transaction.mst_pizza_type AS pt ON p.pizza_type_id = pt.pizza_type_id
GROUP BY
    nama_pizza
ORDER BY
    total_kuantitas_terjual DESC
LIMIT 5
````

### **Analisis: Komplain Paling Sering Terjadi**

Untuk mengidentifikasi komplain yang paling sering terjadi, kita akan menghitung jumlah kemunculan setiap jenis `complain_detail` dari tabel `trx_order` dan mengurutkannya dari yang terbanyak.

````python
%%bigquery --project {project_id} --verbose

SELECT
    complain_detail,
    COUNT(*) AS jumlah_komplain
FROM
    pizza_transaction.trx_order
WHERE
    is_complain = 1 AND complain_detail IS NOT NULL
GROUP BY
    complain_detail
ORDER BY
    jumlah_komplain DESC
LIMIT 5
````

### **Analisis: Jam-Jam Sibuk (Peak Hours)**

Untuk mengidentifikasi jam-jam sibuk, kita akan menghitung jumlah transaksi yang terjadi pada setiap jam dalam sehari.

````python
%%bigquery --project {project_id} --verbose

SELECT
    FORMAT_TIME('%H', PARSE_TIME('%H:%M:%S', order_time)) AS jam_pesanan,
    COUNT(order_id) AS jumlah_transaksi
FROM
    pizza_transaction.trx_order
GROUP BY
    jam_pesanan
ORDER BY
    jumlah_transaksi DESC
LIMIT 10
````
